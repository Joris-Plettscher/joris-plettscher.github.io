### Paper-and-pencil questions: Factorial RM-ANOVA

#### Question 1

In a visual search experiment, **24 participants** search for a target under every combination of

- **Similarity** of target and distractors: Low, High
- **Set size** (number of items on screen): 4, 8, 16

The DV is mean RT (ms). Part of the output is given below.

| Effect | $SS_{effect}$ | $SS_{error}$ |
|---|---:|---:|
| Similarity | 12,000 | 23,000 |
| Set size | 30,000 | 46,000 |
| Similarity × Set size | 4,000 | 46,000 |

1. How many cells does each participant contribute? How many rows does the long-format data set have?
2. Give the effect df and error df for each of the three effects.
3. Compute the F-ratio for each effect. Use $F_{crit}(1,23)=4.28$ and $F_{crit}(2,46)=3.20$ to decide which effects are significant.
4. Compute partial $\eta^2$ for each effect.
5. For which effects must sphericity be checked? Why not for the others?
6. Why does each effect in a factorial RM-ANOVA have **its own** error term, whereas in a two-way independent ANOVA all F-ratios share $MS_E$?

### Answer

#### Question 1

1. $2\times3 = 6$ cells per participant → $24\cdot6=144$ rows in long format.

2. **Degrees of freedom** ($n=24$):

   | Effect | effect df | error df |
   |---|---:|---:|
   | Similarity | $2-1=1$ | $(2-1)(24-1)=23$ |
   | Set size | $3-1=2$ | $(3-1)(24-1)=46$ |
   | Similarity × Set size | $(1)(2)=2$ | $(1)(2)(23)=46$ |

3. **F-ratios** ($F = \dfrac{SS_{effect}/df_{effect}}{SS_{error}/df_{error}}$):

   $$
   F_{Sim}=\frac{12000/1}{23000/23}=\frac{12000}{1000}=12.0 > 4.28 \;\Rightarrow\; \text{significant } (p\approx.002)
   $$

   $$
   F_{Size}=\frac{30000/2}{46000/46}=\frac{15000}{1000}=15.0 > 3.20 \;\Rightarrow\; \text{significant } (p<.001)
   $$

   $$
   F_{Sim\times Size}=\frac{4000/2}{46000/46}=\frac{2000}{1000}=2.0 < 3.20 \;\Rightarrow\; \text{not significant } (p\approx.15)
   $$

4. **Partial $\eta^2 = SS_{effect}/(SS_{effect}+SS_{error})$:**

   - Similarity: $12000/35000=.34$
   - Set size: $30000/76000=.39$
   - Interaction: $4000/50000=.08$

5. Sphericity must be checked (Mauchly) for **Set size** (3 levels) and the **Similarity × Set size** interaction (involves the 3-level factor). Similarity has only 2 levels → only one difference score → sphericity is met automatically.

6. In an RM design every participant contributes to every effect, so each effect has its own **effect × subject** interaction: how consistently the effect appears across participants. This residual is the appropriate error term for that effect. In a between-subjects design there is only one source of error (variation of different people within cells), so all effects share $MS_E$.

### Perceptual Decision-Making Task: Visual Search

### Overview
Participants search a display for a target letter **T** among distractor letters and press a key as soon as they find it. Two factors are manipulated **within subjects**:

| Factor | Levels | Description |
|---|---|---|
| Similarity | Low, High | Distractors are dissimilar (O's) or similar (L's) to the target |
| SetSize | 4, 8, 16 | Number of items on the display |

According to **feature integration theory**, a target that differs clearly from the distractors "pops out" (RT barely depends on set size), whereas a target similar to the distractors needs **serial search** (RT increases steeply with set size).

**Research question:** Do similarity and set size affect RT, and does the set-size effect depend on similarity?

### Step 0: Generate the Data

Run the cell below. It simulates the experiment (24 participants, 20 trials per cell) and saves a trial-level file `VisualSearch_trials.csv`. You don't need to understand this code.

```python
# --- Data simulation: just run this cell ---
import numpy as np
import pandas as pd

rng = np.random.default_rng(3)
rows = []
for s in range(1, 25):
    base = rng.normal(720, 110)      # individual baseline speed
    eff = rng.normal(1, 0.45)        # individual search efficiency
    for sim, slope, shift in [("Low", 4, 0), ("High", 11, 25)]:
        for size in [4, 8, 16]:
            for t in range(20):
                rt = base + shift + eff * slope * size + rng.normal(0, 160)
                rows.append(dict(Participant=f"P{s:02d}", Similarity=sim, SetSize=size,
                                 Trial=t + 1, Correct=rng.random() > (0.04 + 0.002 * size),
                                 RT_ms=rt))
pd.DataFrame(rows).to_csv("VisualSearch_trials.csv", index=False)
print("Saved VisualSearch_trials.csv")
```

### Step 7: Partial $\eta^2$ by Hand

Pingouin does not print the error SS for each effect in a two-way RM-ANOVA. Recover it from F and df, then compute partial $\eta^2$ yourself:

$$
MS_{error} = \frac{MS_{effect}}{F},\qquad SS_{error}=MS_{error}\cdot df_{error},\qquad \eta^2_p=\frac{SS_{effect}}{SS_{effect}+SS_{error}}
$$

Compare with the `np2` column.

```python
tab = anova[["Source", "SS", "ddof1", "ddof2", "MS", "F", "np2"]].copy()

tab["SS_error"] = ___
tab["np2_by_hand"] = ___
tab
```

#### ✅ Solution

```python
tab = anova[["Source", "SS", "ddof1", "ddof2", "MS", "F", "np2"]].copy()

tab["SS_error"] = tab["MS"] / tab["F"] * tab["ddof2"]
tab["np2_by_hand"] = tab["SS"] / (tab["SS"] + tab["SS_error"])
tab.round(3)
```

The hand-computed values match `np2`. Note that each effect has a **different** $SS_{error}$: each effect has its own error term.

#### ✅ Solution

```python
for level in ["Low", "High"]:
    sub = agg[agg["Similarity"] == level]
    res = pg.rm_anova(data=sub, dv="RT_ms", within="SetSize", subject="Participant",
                      correction=True, effsize="np2")
    r = res.iloc[0]
    p_report = r.p_GG_corr if not r.sphericity else r.p_unc
    print(f"{level:>4} similarity: F({r.ddof1}, {r.ddof2}) = {r.F:6.2f}, eps = {r.eps:.2f}, "
          f"sphericity = {r.sphericity}, p = {p_report:.2g}, p_bonf = {min(1, 2 * p_report):.2g}, np2 = {r.np2:.2f}")

posthoc = pg.pairwise_tests(data=agg, dv="RT_ms", within=["Similarity", "SetSize"],
                            subject="Participant", padjust="bonf")
posthoc[posthoc["Contrast"] == "Similarity * SetSize"][["Similarity", "A", "B", "T", "p_unc", "p_corr", "hedges"]].round(4)
```

- **Low similarity:** significant but modest set-size effect, $F(2,46)=7.05,\ p_{bonf}=.004,\ \eta^2_p=.23$ (≈ 751 → 797 ms).
- **High similarity:** much larger set-size effect, $F(2,46)=34.76,\ p_{bonf}<.001,\ \eta^2_p=.60$ (≈ 800 → 933 ms).
- Sphericity holds within each level (Mauchly n.s.), so uncorrected df are reported for the simple effects.
- Pairwise tests (Bonferroni over the 6 within-level comparisons): for **High** similarity all three set sizes differ (4 vs 8: $p_{bonf}=.033$; others $p<.001$). For **Low** similarity only **4 vs 16** differs ($p_{bonf}=.008$); neighbouring set sizes do not.
- Interpretation: the interaction reflects a **steeper search slope** for similar distractors (more serial search), in line with feature integration theory. A plain "RT increases with set size" (main effect) hides this difference.

### Finally:

**Question:** Do we need post hoc tests for each effect in this design? Why or why not?

### Answer

- **Similarity:** no. It has only two levels, so the F-test already compares the two means.
- **SetSize:** yes. Three levels: the omnibus F only says that *at least one* set size differs. Pairwise comparisons are needed (Bonferroni here, because sphericity is violated for SetSize).
- **Interaction:** yes. A significant interaction should be followed up with **simple effects** and an interaction plot to show *where* the conditions differ.
